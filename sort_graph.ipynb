# **vẽ line graph cho thời gian thực thi của các hàm sort trong chương trình chạy bằng C++**

```python
import matplotlib.pyplot as plt

files = [
    'test_1.txt', 'test_2.txt', 'test_3.txt', 'test_4.txt', 'test_5.txt',
    'test_6.txt', 'test_7.txt', 'test_8.txt', 'test_9.txt', 'test_10.txt'
]

quicksort = [76.03, 80.23, 213.99, 220.14, 248.97, 229.82, 295.15, 218.59, 221.82, 216.13]
heapsort  = [496.79, 476.99, 609.19, 595.07, 619.98, 631.96, 718.94, 643.48, 611.49, 579.86]
mergesort = [352.43, 336.77, 479.06, 486.03, 535.10, 496.55, 496.50, 491.52, 464.72, 481.59]
std_sort  = [48.67, 39.17, 129.15, 130.55, 135.01, 132.66, 130.69, 136.74, 139.98, 141.55]

plt.figure(figsize=(10, 6))

plt.plot(files, quicksort, marker='o', linewidth=2, color='#1f77b4', label='QuickSort')
plt.plot(files, heapsort,  marker='s', linewidth=2, color='#d62728', label='HeapSort')
plt.plot(files, mergesort, marker='^', linewidth=2, color='#ff7f0e', label='MergeSort')
plt.plot(files, std_sort,  marker='d', linewidth=2, color='#2ca02c', label='std::sort')

plt.title('Tốc độ thực thi của các thuật toán sắp xếp (N = 1.000.000)', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('Bộ dữ liệu (Test Files)', fontsize=11)
plt.ylabel('Thời gian thực thi (ms)', fontsize=11)

plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Thuật toán', fontsize=10, title_fontsize=11)
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig('sorting_line_chart.png', dpi=300)
plt.show()
```

# **thời gian trung bình, nhanh nhất, chậm nhất của các thuật toán sort trong C++**

```python
import matplotlib.pyplot as plt
import numpy as np

metrics = ['Nhanh nhat (Min)', 'Trung binh (Avg)', 'Cham nhat (Max)']
algorithms = ['std::sort', 'QuickSort', 'MergeSort', 'HeapSort']

data = {
    'std::sort': [39.17, 116.42, 141.55],
    'QuickSort': [76.03, 202.09, 295.15],
    'MergeSort': [336.77, 462.03, 535.10],
    'HeapSort': [476.99, 598.38, 718.94]
}

colors = {
    'std::sort': '#2ca02c',
    'QuickSort': '#1f77b4',
    'MergeSort': '#ff7f0e',
    'HeapSort': '#d62728'
}

x = np.arange(len(metrics))
width = 0.18

fig, ax = plt.subplots(figsize=(10, 6))

for i, alg in enumerate(algorithms):
    offset = (i - 1.5) * width
    rects = ax.bar(x + offset, data[alg], width, label=alg, color=colors[alg], edgecolor='white', linewidth=0.8)

    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.1f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 4),
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=9)

ax.set_title('So sánh thời gian: Nhanh nhất - Trung bình - Chậm nhất (ms)', fontsize=13, fontweight='bold', pad=15)
ax.set_xlabel('Chỉ số thống kê', fontsize=11, labelpad=8)
ax.set_ylabel('Thời gian thực thi (ms)', fontsize=11)
ax.set_xticks(x)
ax.set_xticklabels(['Nhanh nhất (Min)', 'Trung bình (Avg)', 'Chậm nhất (Max)'], fontsize=10)
ax.set_ylim(0, 800)
ax.grid(True, axis='y', linestyle='--', alpha=0.6)
ax.legend(title='Thuật toán', fontsize=10, title_fontsize=11)

plt.tight_layout()

plt.savefig('summary_statistics_chart.png', dpi=300)
plt.show()
```
